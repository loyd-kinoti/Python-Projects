```python
Name = input( "Enter your name:" )
weight=int(input("Enter your weight:"))
Height=int(input("Enter your Height in inches:"))
BMI = (weight * 703) / (Height * Height )
print ("Your BMI is:",BMI)
if BMI>0:
    if(BMI<18.5):
        print ("you're Underweight and has minimal healthrisk")
    elif (BMI<24.9):
        print ("you're normal weight and has minimal healthrisk")
    elif (BMI<29.9):
        print ("you're overweight and has increased healthrisk")
    elif (BMI<34.9):
        print ("you're Obese and has high healthrisk")
    elif (BMI<39.9):
        print ("you're severely obese and has very high healthrisk")
    elif (BMI>40):
        print ("you're morbidly obese and has extremely high healthrisk")
else:
    print("Enter valid inputs")
```
